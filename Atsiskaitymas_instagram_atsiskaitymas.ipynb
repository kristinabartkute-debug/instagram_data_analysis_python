## 13. Sekėjų lygiai

Suskirstykite įrašus pagal autoriaus sekėjų skaičių: Nano (iki 1 000), Micro (1 000–9 999), Mid (10 000–99 999), Macro (100 000 ir daugiau). Kiekvienam lygiui pateikite unikalių paskyrų skaičių, įrašų skaičių ir vidutinį pasiekiamumą. Lygius pateikite nuo mažiausio iki didžiausio.

**Pavyzdinis rezultatas**

| tier | paskyru_sk | irasu_sk | vid_reach |
| --- | --- | --- | --- |
| Nano | 72 | 1 348 | 456 |
| Micro | 159 | 2 986 | 2 920 |
| Mid | 82 | 1 529 | 22 324 |
| Macro | 7 | 137 | 111 858 |

```python
def sekejai(eilute):
  if eilute<1000:
    return 'Nano'
  elif eilute>=1000 and eilute <=9999:
    return 'Micro'
  elif eilute>=10000 and eilute<=99999:
    return 'Mid'
  else:
    return 'Macro'
data['tier']=data['followers'].apply(sekejai)
r=data.groupby('tier').agg(
    paskyru_sk= ('username','nunique'),
    irasu_sk=('post_id', 'count'),
    vid_reach=('reach', 'mean')
).round(0).astype(int)

lygiai=['Nano', 'Micro', 'Mid','Macro']
r=r.reindex(lygiai)
r
```

## 21. Aprašymo ilgis ir komentarai

Suskirstykite įrašus pagal aprašymo ilgį: Trumpas (iki 100 simbolių), Vidutinis (100–300), Ilgas (daugiau nei 300). Pateikite lentelę, kurioje eilutės yra aprašymo grupės, stulpeliai yra įrašų tipai, o reikšmės yra vidutinis komentarų skaičius.

**Pavyzdinis rezultatas**

| caption_group | Carousel | Photo | Reel | Video |
| --- | --- | --- | --- | --- |
| Trumpas | 10.8 | 10.7 | 28.1 | 10.6 |
| Vidutinis | 14.6 | 8.2 | 40.9 | 10.2 |
| Ilgas | 14.5 | 6.3 | 32.3 | 5.2 |

```python
def ilgis(eilute):
  if eilute<100:
    return 'Trumpas'
  elif eilute>=100 and eilute<300:
    return 'Vidutinis'
  elif eilute>=300:
    return 'Ilgas'
data['caption_group']=data['caption_length'].apply(ilgis)
d=data.pivot_table(
    index='caption_group',
    columns='post_type',
    values='comments',
    aggfunc='mean'
)
d=d.reindex(['Trumpas', 'Vidutinis', 'Ilgas'])
d.round(1)
```

## 27. Grotažymių kiekis ir pasiekiamumas

Apskaičiuokite, kokią dalį sekėjų pasiekė kiekvienas įrašas (pasiekiamumas padalintas iš sekėjų). Suskirstykite įrašus pagal grotažymių kiekį: 0–3, 4–6, 7–9, 10 ir daugiau. Kiekvienai grupei pateikite įrašų skaičių ir šio santykio medianą. Business tipo paskyrų neįtraukite.

**Pavyzdinis rezultatas**

| hashtag_group | irasu_sk | mediana |
| --- | --- | --- |
| 0–3 | 707 | 0.577 |
| 4–6 | 2 123 | 0.563 |
| 7–9 | 1 496 | 0.572 |
| 10+ | 423 | 0.569 |

```python
def hash(eilute):
  if eilute<=3:
    return '0-3'
  elif eilute>=4 and eilute<=6:
    return '4-6'
  elif eilute>=7 and eilute<=9:
    return '7-9'
  else:
    return '10+'
data['sekeju_dalis']=data['reach']/data['followers']
data['hashtag_group']=data['hashtags_count'].apply(hash)
d=data.loc[data['account_type']!='Business'].groupby('hashtag_group').agg(
    irasu_sk=('post_id','count'),
    mediana=('sekeju_dalis', 'median')
).round(3)
d=d.reindex(['0-3', '4-6','7-9','10+'])
d
```
